```python
import numpy as np
from bokeh.plotting import figure, output_file, show
from bokeh.io import output_notebook
from bokeh.models import Slider, CustomJS, ColumnDataSource
from bokeh.layouts import column
```

```python


output_notebook()

# Set up data
x = np.linspace(0, 2 * np.pi, 500)
y = np.sin(x)

source = ColumnDataSource(data=dict(x=[x[0]], y=[y[0]]))

# Create a plot
plot = figure(height=400, width=400)
plot.line(x, y, line_width=2, alpha=0.8)
point = plot.scatter('x', 'y', size=10, source=source)  # Updated method

# Set up the slider
slider = Slider(start=0, end=2 * np.pi, value=0, step=0.01, title="Slider")

# Define the callback function for the slider
callback = CustomJS(args=dict(source=source, x=x, y=y), code="""
    const data = source.data;
    const f = cb_obj.value;
    data['x'] = [f];
    data['y'] = [Math.sin(f)];
    source.change.emit();
""")

slider.js_on_change('value', callback)

layout = column(plot, slider)

show(layout)
```
